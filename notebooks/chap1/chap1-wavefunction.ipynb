```python
import numpy as np
import matplotlib.pyplot as plt
import scienceplots
```

```python
def style_ax(ax, xlabel='$x$', ylabel='$y$', xlim=None, ylim=None,
             xticks=None, xticklabels=None, yticks=None, yticklabels=None,
             color='white', fontsize=16):
    """Apply shared plot attributes to one or more axes."""
    axes = np.atleast_1d(ax)
    for a in axes:
        a.set_xlabel(xlabel, color=color, fontsize=fontsize)
        a.set_ylabel(ylabel, color=color, fontsize=fontsize)
        a.tick_params(colors=color)
        a.tick_params(labelsize=fontsize)
        a.tick_params(which='minor', top=False, bottom=False, left=False, right=False)
        a.grid(False)
        if xlim is not None:
            a.set_xlim(xlim)
        if ylim is not None:
            a.set_ylim(ylim)
        if xticks is not None:
            a.set_xticks(xticks)
        if xticklabels is not None:
            a.set_xticklabels(xticklabels, color=color)
        if yticks is not None:
            a.set_yticks(yticks)
        if yticklabels is not None:
            a.set_yticklabels(yticklabels, color=color)
```

```python
# rho(x) = A exp(-lambda (x - a)^2),  A = sqrt(lambda/pi),  sigma = 1/sqrt(2 lambda)
# Gaussian centered at a, peak A, inflection points at a +/- sigma. Take a = 0, lambda = 1
# and label ticks symbolically so the sketch holds for any a, lambda.
a, lam = 0.0, 1.0
A = np.sqrt(lam/np.pi)
sigma = 1/np.sqrt(2*lam)

x = np.linspace(a - 4*sigma, a + 4*sigma, 801)
rho = A*np.exp(-lam*(x - a)**2)
rho_sigma = A*np.exp(-0.5)   # value at the inflection points

plt.style.use(['science', 'scatter', 'dark_background'])
fig, ax = plt.subplots(figsize=(8, 6))

ax.plot(x, rho, color="#00F5FF", linewidth=2.5, label=r'$\rho(x) = \sqrt{\lambda/\pi}\, e^{-\lambda (x - a)^2}$')
ax.vlines([a - sigma, a, a + sigma], 0, [rho_sigma, A, rho_sigma],
          colors='white', linestyles='--', linewidths=1.2)
ax.plot([a - sigma, a + sigma], [rho_sigma, rho_sigma], 'o', color="#FF073A", markersize=8,
        zorder=3, label=r'inflection points $x = a \pm \sigma$')
ax.plot(a, A, 'o', color='white', markersize=8, zorder=3, label=r'peak $\rho(a) = \sqrt{\lambda/\pi}$')

style_ax(ax, xlabel='$x$', ylabel=r'$\rho(x)$', xlim=(a - 4*sigma, a + 4*sigma), ylim=(0, 1.15*A),
         xticks=[a - sigma, a, a + sigma], xticklabels=[r'$a - \sigma$', '$a$', r'$a + \sigma$'],
         yticks=[0, rho_sigma, A], yticklabels=['$0$', r'$\sqrt{\lambda/\pi e}$', r'$\sqrt{\lambda/\pi}$'])
ax.legend(frameon=True, fontsize=14, loc='lower center', bbox_to_anchor=(0.5, 1.02), ncol=1)

fig.savefig("chap1ex1.3c.pdf", format='pdf', bbox_inches='tight')
plt.show()
```
